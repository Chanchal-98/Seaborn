```python
# Task 1 Relational Plot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_data.csv")

## Relational plot age and Absences
sns.relplot(data=df,x='age',y='absences', hue='sex', kind='scatter')
plt.show()
```

```python
## Task 2 Line Plot as scatter $ facet
sns.relplot(data=df, x='age', y='famrel',kind='line')

# scatter-style line plot
sns.relplot(data=df, x='age', y='famrel', marker='o', linestyle='--', kind='line')
plt.show()

## plot based on a categorical column
sns.relplot(data=df,x='famrel',y='age',col='sex',kind='scatter')
plt.show() 
```

```python
## Task 3 Distribution Plots

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_data.csv")

# Histogram Plot
sns.displot(data=df,x='Fedu',y='age',kind='hist')

## KDE Plot
sns.kdeplot(data=df,x='age',fill=True)

## Rug Plot
sns.rugplot(data=df,x='age')
plt.show()


## Histogram and KDE Plot
sns.histplot(data=df, x="age", kde=True)
```

```python
## Task 4 Bivariate Distribution Plots

# Bivariate histogram
sns.histplot(data=df,x='Medu',y='Fedu')

# Bivariate KDE Plot
sns.kdeplot(data=df,x='Medu',y='Fedu')
```

```python
## Task 5 Matrix Plot

# Pair plot
sns.pairplot(df[['age', 'Medu', 'Fedu']])

# Heatmap of correlation Matrix
dt = df.pivot_table(index='Fjob',columns='freetime', values='goout')
sns.heatmap(data=dt)
```

```python
## Task 6 Categorical Plot

## Bar plot
sns.barplot(data=df,x='sex',y='age')

## Box Plot
sns.boxplot(data=df, x='sex',y='age')

## Count plot 
sns.countplot(data=df,x='Mjob')
```

```python
# violin plot
sns.violinplot(data=df,x='freetime')
```

```python
## Task 7 Regressionn Plots

##  Regression Plot
sns.regplot(data=df,x='Medu',y='Fedu')


```

```python
## Implot with uhue using a categorical plot
sns.scatterplot(data=df, x="age", y="goout", hue="sex")
```

```python
## Task 8 Multi plots and figure level plots

## FacetGrid plot
sns.FacetGrid(
    data=df,
    row="sex",
    col="age"
)
```

```python
## Multiplot dashboard

#1. relplot
sns.relplot(data=df, x="age", y="freetime", hue="sex")

#2. catplot
sns.catplot(data=df, x="Mjob", y="freetime", kind="box")

#3. displot
sns.displot(data=df, x="age", kde=True)
```
